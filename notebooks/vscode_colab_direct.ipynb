```python
# Import E1 experiment and run directly
import sys
sys.path.insert(0, '/content/drive/My Drive/Panini') if 'google.colab' in sys.modules else None

from pathlib import Path
import json

# Option 1: Clone from GitHub directly
import subprocess
repo_path = Path('/content/panini-vscode')
if not repo_path.exists():
    subprocess.run(['git', 'clone', 'https://github.com/stephanedenis/Panini.git', str(repo_path)], 
                   cwd='/content')

sys.path.insert(0, str(repo_path))
print(f"✅ Panini repo ready at {repo_path}")
```

```python
# Run E1 experiment directly
import subprocess
result = subprocess.run(
    ['python3', str(repo_path / 'experiments' / 'e1_format_decomposition.py'), '--phase', 'all'],
    cwd=str(repo_path),
    capture_output=True,
    text=True
)

print(result.stdout)
if result.returncode != 0:
    print(f"❌ Error: {result.stderr}")
else:
    print(f"\n✅ E1 execution completed successfully")
```

```python
# Load and display results
results_file = repo_path / 'outputs' / 'e1_results.json'
if results_file.exists():
    with open(results_file) as f:
        results = json.load(f)
    
    print(f"Experiment: {results['experiment']}")
    print(f"Hypothesis: {results['hypothesis']}")
    print(f"Overall Status: {results['overall_status']}")
    print(f"\nPhase Results:")
    for phase, data in results['phases'].items():
        print(f"  {phase}: {data.get('status', 'unknown').upper()}")
else:
    print(f"Results file not found at {results_file}")
```
